```python
def calc_notas ():
  notas1 = [7, 8, 9]
  nota = 0
  soma = 0

  for nota in notas1:
    soma = soma + nota
    media = soma / len(notas1)
  print (f"a soma das notas é {soma}")
  print (f"a média das notas é {media}")

x = int(input("Quer chamar função de calc notas? Se sim digite 1  "))

if x == 1:
  calc_notas()
else:
  print("você não quiser ver as notas")
```
