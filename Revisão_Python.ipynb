Crie um programa que pergunte ao usuário um número e, em seguida, calcule e mostre na tela a tabuada de multiplicação desse número, do 1 ao 10.

```python
numero = int(input("diga um número para fazer a tabuada do 1 a 10  "))

for x in range(0, 11, 1):
  print (numero * x)
```
